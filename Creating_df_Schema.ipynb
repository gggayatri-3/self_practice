```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('test').getOrCreate()
```

```python
columns = ['id', 'name']
data = [(1, 'Gayatri'), (2, 'Ram'), (3, 'Anir')]
df = spark.createDataFrame(data, columns)
df.show()
df.printSchema()
```

```python
from pyspark.sql.types import StructField, StructType, IntegerType, StringType
schema = StructType([StructField('id', IntegerType()),
         StructField('name', StringType())])
data = [(1, 'Gayatri'), (2, 'Ram'), (None, None), (3, 'Anir'), (4, 'Anir')]
df = spark.createDataFrame(data, schema =schema)
df.show()
```

```python
# Do Not Accept The Null Values(Default is True - can accept Null Values as well)
from pyspark.sql.types import StructField, StructType, IntegerType, StringType
schema = StructType([StructField('id', IntegerType(), False),
         StructField('name', StringType(), False)])
data = [(1, 'Gayatri'), (2, 'Ram'), (5, 'Rohan'), (3, 'Anir'), (4, 'Anir')]
df = spark.createDataFrame(data, schema =schema)
df.show()
```

```python
from IPython.utils.ipstruct import Struct
from datetime import date
from pyspark.sql.types import StructField, StructType, IntegerType, StringType, BooleanType, DateType, DoubleType
emp_schema = StructType([StructField('id', IntegerType(), True),
                     StructField('name', StringType(), True),
                     StructField('Date Of Joining', DateType(), True),
                     StructField('Salary', DoubleType(), True),
                     StructField('Status', BooleanType(), True)])
data = [(1,'Gayatri', date(2023,11,23), 70000.00, True),
        (2,'Anirban', date(2024,7,23), 70500.23, True),
        (3,'Ram', date(2025,11,11), 45000.40, True),
        (4,'Rahul', date(2025,11,13), 34000.40, False),
        (5,'Sima', date(2025,11,22), 20000.00, True)]
df = spark.createDataFrame(data, schema = emp_schema)
df.show()
```
