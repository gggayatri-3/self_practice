```python
# parallelize () helps us convert any list or tuple into an RDD
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("sample_RDD").getOrCreate()
number = [10,20,30,40,50]

sc = spark.sparkContext

rdd = sc.parallelize(number)
print(type(rdd))
print(rdd.collect())
```

```python
# Multiply eace value in RDD with 5
rdd2 =rdd.map(lambda x: x*5)
print(rdd2.collect())

rdd2 = rdd.map(lambda i: i + 5)
print(rdd2.collect())

# create new RDD with value >= 40
rdd3 = rdd.filter(lambda x: x >= 40)
print(rdd3.collect())
```

```python
strings = ['hello','students','pyspark']
rdd1 = sc.parallelize(strings)
upp_rdd = rdd1.map(lambda x: x.upper())
len_rdd = rdd1.map(lambda x: len(x))
print(upp_rdd.count())
print(len_rdd.collect())
print(upp_rdd.collect())
```

```python
l1= ['Hello students, how are you? pyspark session started']
rdd1 = sc.parallelize(l1)
split_rdd = rdd1.map(lambda x: x.split(' '))
print(split_rdd.count())
print(split_rdd.collect())
```

```python
# finding the length of each string element using RDD
# flatMap - extension of map function
# map - applies the operation on each element
# flatMap - applies the operation on every element and makes each element individual
# individual element creation - flattens the data
l1= ['Hello students, how are you? pyspark session started']
rdd1 = sc.parallelize(l1)
split_count_rdd = rdd1.flatMap(lambda x: x.split(" "))
print(split_count_rdd.count())
print(split_count_rdd.collect())
```

```python
from posixpath import split
l1= [['Hello students,'], ['how are you?'], ['pyspark session'], ['started']]
rdd1 = sc.parallelize(l1)
split_list_count_rdd = rdd1.flatMap(lambda x: x[0].split(' '))
print(split_list_count_rdd.count())
print(split_list_count_rdd.collect())
```

```python
# Displaying first 3 elements from rdd list
print(split_list_count_rdd.take(3))
```

```python
# printing 1st value from the list
print(split_list_count_rdd.first())
```

```python
# print how many times each word is present in an RDD
l2 = [['hi students'], ['hi hello you?'], ['pyspark session'], ['pyspark']]
rdd1 = sc.parallelize(l2)
seperate_rdd = rdd1.flatMap(lambda x: x[0].split(' '))
print(seperate_rdd.collect())
```

```python
map_rdd = seperate_rdd.map(lambda x: (x,1))
red_rdd = map_rdd.reduceByKey(lambda a,b : a+b)
print(red_rdd.collect())
grp_rdd = map_rdd.groupBy(lambda word:word)
res_rdd = grp_rdd.mapValues(len)
print(res_rdd.collect())
```

```python
l = [10,20,30,40,50,60,70,80]
num_rdd = sc.parallelize(l)
print(num_rdd.collect())
sum_rdd = num_rdd.reduce(lambda a,b : a+b)
print(sum_rdd)

```

**1. map flatmap function difference**
--> map() gives exactly ONE output for each input.
flatMap() can give ZERO, ONE, or MANY outputs for each input, and then flattens them.


**2. how does collect() works in pyspark RDD?**
-->collect() brings all the data from the RDD's worker nodes back to the Driver program as a Python list.
its job is to bring the complete RDD result from the distributed workers back to the Driver.


**3. Difference between reduceByKey and GroupByKey**
--> reduceByKey() combines values while shuffling, whereas groupByKey() collects all values for each key without combining them first.
reduceByKey() performs aggregation on values with the same key and uses local combining to reduce shuffle data, making it generally more efficient. groupByKey() simply groups all values associated with each key and does not perform local aggregation, so it can require more memory and network traffic.


**4. RDD inbuilt functions**
--> The main RDD inbuilt functions are map(), flatMap(), filter(), distinct(), union(), intersection(), groupByKey(), reduceByKey(), collect(), count(), first(), take(), reduce(), max(), min(), and sum().
